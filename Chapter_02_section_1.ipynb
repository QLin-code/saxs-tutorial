## SLD contrast for silica spheres

The scattering contrast is calculated as:

\[
\Delta \rho = 18.8 \times 10^{-6} - 9.47 \times 10^{-6}
= 9.33 \times 10^{-6}\ \text{Å}^{-2}
\]

For silica spheres, the particle SLD is \(18.8 \times 10^{-6}\ \text{Å}^{-2}\). I will compare the contrast in pure water and heavy water.

```python
rho_silica = 18.8e-6
rho_water = 9.47e-6
rho_heavy_water = 6.40e-6

delta_rho_water = rho_silica - rho_water
delta_rho_heavy_water = rho_silica - rho_heavy_water

print(f"Delta rho in pure water: {delta_rho_water:.2e} Å^-2")
print(f"Delta rho in heavy water: {delta_rho_heavy_water:.2e} Å^-2")
```

### Answers

1. For silica spheres in pure water:

\[
\Delta \rho = 18.8 \times 10^{-6} - 9.47 \times 10^{-6}
= 9.33 \times 10^{-6}\ \text{Å}^{-2}
\]

2. The contrast increases in heavy water. In heavy water:

\[
\Delta \rho = 18.8 \times 10^{-6} - 6.40 \times 10^{-6}
= 12.40 \times 10^{-6}\ \text{Å}^{-2}
\]

This is larger than the contrast in pure water.

3. No, it is not possible to reach \(\Delta \rho = 0\) for silica using only this water/heavy-water mixture. To reach zero contrast, the solvent would need to have the same SLD as silica:

\[
\rho_{\text{solvent}} = 18.8 \times 10^{-6}\ \text{Å}^{-2}
\]

However, mixtures of water and heavy water can only have SLD values between  
\(6.40 \times 10^{-6}\) and \(9.47 \times 10^{-6}\ \text{Å}^{-2}\), which is much lower than silica.

## SLD and optical constants conversion utilities

The following functions convert between optical constants \(\delta, \beta\) and complex SLD components.

```python
import numpy as np

HC = 12398.4  # hc in eV·Å — converts photon energy to wavelength

def wavelength_from_energy(energy_eV):
    """Return X-ray wavelength in Å for a given photon energy in eV."""
    return HC / energy_eV

def sld_from_optical_constants(delta, beta, energy_eV):
    """
    Convert optical constants (delta, beta) to complex SLD.

    Parameters
    ----------
    delta : float
        Refractive index decrement (dimensionless).
    beta : float
        Absorption index (dimensionless).
    energy_eV : float
        Photon energy in eV.

    Returns
    -------
    rho_real, rho_imag : float
        Real and imaginary SLD components in Å^-2.
    """
    lam = wavelength_from_energy(energy_eV)
    prefactor = 2 * np.pi / lam**2
    return prefactor * delta, prefactor * beta

def optical_constants_from_sld(rho_real, rho_imag, energy_eV):
    """
    Convert complex SLD to optical constants (delta, beta).

    Parameters
    ----------
    rho_real : float
        Real SLD component in Å^-2.
    rho_imag : float
        Imaginary SLD component in Å^-2.
    energy_eV : float
        Photon energy in eV.

    Returns
    -------
    delta, beta : float
        Dimensionless optical constants.
    """
    lam = wavelength_from_energy(energy_eV)
    prefactor = lam**2 / (2 * np.pi)
    return prefactor * rho_real, prefactor * rho_imag
```

```python
# Typical optical constants for a polymer near the carbon K-edge
energy = 285.0   # eV — near the carbon 1s absorption edge
delta  = 2.5e-4  # dimensionless
beta   = 1.5e-5  # dimensionless

rho_real, rho_imag = sld_from_optical_constants(delta, beta, energy)
lam = wavelength_from_energy(energy)

print(f"Photon energy:  {energy:.1f} eV")
print(f"Wavelength:     {lam:.4f} Å")
print(f"rho_real:       {rho_real:.3e} Å^-2")
print(f"rho_imag:       {rho_imag:.3e} Å^-2")
```
