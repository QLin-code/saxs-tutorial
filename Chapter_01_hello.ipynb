```python
import numpy as np
import matplotlib.pyplot as plt

# 200 evenly spaced values from 0 to 2*pi
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y, color="steelblue", label=r"$\sin(x)$")
ax.set_xlabel(r"$x$ (radians)")
ax.set_ylabel(r"$\sin(x)$")
ax.legend(loc="upper right")
plt.show()
```

The x-axis represents the angle x in radians from 0 to 2π. The y-axis represents sin(x), which ranges from -1 to 1. One full period of the sine wave goes from 0 to 2π.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
y_sin = np.sin(x)
y_cos = np.cos(x)

fig, ax = plt.subplots()
ax.plot(x, y_sin, color="steelblue", label=r"$\sin(x)$")
ax.plot(x, y_cos, color="orange", label=r"$\cos(x)$")
ax.set_xlabel(r"$x$ (radians)")
ax.set_ylabel("value")
ax.legend(loc="upper right")
plt.show()
```
